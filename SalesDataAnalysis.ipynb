```python
# Step 1: Import only Pandas and NumPy
import pandas as pd
import numpy as np

# Set clean display format for numbers
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.float_format', lambda x: '%.2f' % x)

print("Libraries imported successfully!")
print(f"Pandas Version: {pd.__version__}")
print(f"NumPy Version : {np.__version__}")
```

```python
# Step 2: Load the raw dataset from CSV
# Place 'retail_sales_data.csv' in your Colab files
df = pd.read_csv('sampledata1.csv')

print("--- Dataset Loaded Successfully ---")
print(f"Total Rows (Records)    : {df.shape[0]}")
print(f"Total Columns (Features): {df.shape[1]}")
print("\nFirst 5 rows of data:")
print(df.head())

print("\n--- Column Names & Data Types ---")
print(df.info())

print("\n--- Missing Values Check ---")
print(df.isnull().sum())
```

```python
# Step 3: Executive Sales Summary (Key Performance Indicators)
total_gross = np.round(df['Gross_Sales'].sum(), 2)
total_net = np.round(df['Total_Amount'].sum(), 2)
total_discount = np.round(total_gross - total_net, 2)
total_units = int(df['Quantity'].sum())
total_orders = len(df)
avg_order_value = np.round(df['Total_Amount'].mean(), 2)
avg_discount_pct = np.round(df['Discount_Pct'].mean() * 100, 2)

print("=" * 50)
print("         EXECUTIVE SALES OVERVIEW")
print("=" * 50)
print(f"Total Gross Revenue      : ${total_gross:,.2f}")
print(f"Total Net Revenue (Sales): ${total_net:,.2f}")
print(f"Total Discount Given     : ${total_discount:,.2f}")
print(f"Total Units Sold         : {total_units:,}")
print(f"Total Transactions       : {total_orders:,}")
print(f"Average Order Value (AOV): ${avg_order_value:,.2f}")
print(f"Average Discount Rate    : {avg_discount_pct}%")
print("=" * 50)
```

```python

# Question 1: What is the overall sales value and how does it change over time?


# Group sales by month
monthly_sales = df.groupby(['Month_Num', 'Month'])['Total_Amount'].agg(
    Monthly_Revenue='sum',
    Total_Orders='count',
    Average_Order_Value='mean'
).reset_index()

# Sort chronologically by Month Number
monthly_sales = monthly_sales.sort_values('Month_Num')

# Calculate Month-over-Month (MoM) Growth Percentage
monthly_sales['MoM_Growth_%'] = monthly_sales['Monthly_Revenue'].pct_change() * 100
monthly_sales['MoM_Growth_%'] = monthly_sales['MoM_Growth_%'].fillna(0).round(2)
monthly_sales['Monthly_Revenue'] = monthly_sales['Monthly_Revenue'].round(2)
monthly_sales['Average_Order_Value'] = monthly_sales['Average_Order_Value'].round(2)

print("--- Monthly Sales Trend & MoM Growth ---")
print(monthly_sales[['Month_Num', 'Month', 'Monthly_Revenue', 'Total_Orders', 'Average_Order_Value', 'MoM_Growth_%']])

best_month = monthly_sales.loc[monthly_sales['Monthly_Revenue'].idxmax()]
lowest_month = monthly_sales.loc[monthly_sales['Monthly_Revenue'].idxmin()]

print(f"\nHighest Earning Month: {best_month['Month']} (${best_month['Monthly_Revenue']:,.2f})")
print(f"Lowest Earning Month : {lowest_month['Month']} (${lowest_month['Monthly_Revenue']:,.2f})")
```

```python

# Question 2: Which categories and products contribute the largest and smallest share of revenue?


# 1. Category-level Revenue Share
category_summary = df.groupby('Category').agg(
    Revenue=('Total_Amount', 'sum'),
    Units_Sold=('Quantity', 'sum'),
    Transactions=('Transaction_ID', 'count')
).reset_index()

total_rev = category_summary['Revenue'].sum()
category_summary['Revenue_Share_%'] = (category_summary['Revenue'] / total_rev * 100).round(2)
category_summary = category_summary.sort_values(by='Revenue', ascending=False)

print("--- Category Revenue Share ---")
print(category_summary)

# 2. Product-level Performance
product_summary = df.groupby(['Category', 'Product_Name']).agg(
    Total_Sales=('Total_Amount', 'sum'),
    Units_Sold=('Quantity', 'sum')
).reset_index().sort_values(by='Total_Sales', ascending=False)

top_3_products = product_summary.head(3)
bottom_3_products = product_summary.tail(3)

print("\n--- Top 3 Best-Selling Products (Largest Share) ---")
print(top_3_products)

print("\n--- Bottom 3 Lowest-Selling Products (Smallest Share) ---")
print(bottom_3_products)
```

```python

# Question 3: Which branches perform best and which need further investigation?


branch_perf = df.groupby('Branch').agg(
    Total_Sales=('Total_Amount', 'sum'),
    Total_Orders=('Transaction_ID', 'count'),
    Units_Sold=('Quantity', 'sum'),
    Avg_Order_Value=('Total_Amount', 'mean')
).reset_index()

total_chain_sales = branch_perf['Total_Sales'].sum()
branch_perf['Sales_Share_%'] = (branch_perf['Total_Sales'] / total_chain_sales * 100).round(2)
branch_perf['Avg_Order_Value'] = branch_perf['Avg_Order_Value'].round(2)
branch_perf = branch_perf.sort_values(by='Total_Sales', ascending=False)

print("--- Branch Performance Table ---")
print(branch_perf)

best_branch = branch_perf.iloc[0]
weakest_branch = branch_perf.iloc[-1]

print(f"\nBest Performing Branch : {best_branch['Branch']} with ${best_branch['Total_Sales']:,.2f} ({best_branch['Sales_Share_%']}%)")
print(f"Needs Investigation    : {weakest_branch['Branch']} with ${weakest_branch['Total_Sales']:,.2f} ({weakest_branch['Sales_Share_%']}%)")
```

```python

# Question 4: Which months, weekdays or periods show stronger or weaker sales?


# Weekday Sales Analysis
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

weekday_sales = df.groupby('Day_of_Week').agg(
    Total_Sales=('Total_Amount', 'sum'),
    Order_Count=('Transaction_ID', 'count'),
    Avg_Sale=('Total_Amount', 'mean')
).reindex(day_order).reset_index()

weekday_sales['Sales_Share_%'] = (weekday_sales['Total_Sales'] / weekday_sales['Total_Sales'].sum() * 100).round(2)
weekday_sales['Avg_Sale'] = weekday_sales['Avg_Sale'].round(2)

print("--- Sales by Day of Week (Chronological) ---")
print(weekday_sales)

# Rank days from strongest to weakest
ranked_days = weekday_sales.sort_values(by='Total_Sales', ascending=False)
print("\n--- Weekdays Ranked by Sales ---")
print(ranked_days[['Day_of_Week', 'Total_Sales', 'Sales_Share_%']])

print(f"\nStrongest Day: {ranked_days.iloc[0]['Day_of_Week']} (${ranked_days.iloc[0]['Total_Sales']:,.2f})")
print(f"Weakest Day  : {ranked_days.iloc[-1]['Day_of_Week']} (${ranked_days.iloc[-1]['Total_Sales']:,.2f})")
```

```python

# Question 5: How does the average transaction value (ATV) vary across categories and locations?


# Create a 2D Pivot Table of Average Transaction Value (Branch vs Category)
atv_pivot = df.pivot_table(
    index='Branch',
    columns='Category',
    values='Total_Amount',
    aggfunc='mean'
).round(2)

# Add Branch Overall Average
atv_pivot['Overall_Branch_Avg'] = df.groupby('Branch')['Total_Amount'].mean().round(2)

print("--- Average Transaction Value ($) by Branch & Category ---")
print(atv_pivot)

# Category Overall Average
cat_overall = df.groupby('Category')['Total_Amount'].mean().round(2)
print("\n--- Overall Average Transaction Value by Category ---")
print(cat_overall.sort_values(ascending=False))
```

```python

# Question 6: Are larger discounts associated with higher quantities or transaction values?


# 1. Create Discount Bins (Tiers)
discount_bins = [-0.01, 0.00, 0.05, 0.10, 0.25]
discount_labels = ['No Discount (0%)', 'Low (5%)', 'Medium (10%)', 'High (15-25%)']
df['Discount_Tier'] = pd.cut(df['Discount_Pct'], bins=discount_bins, labels=discount_labels)

discount_impact = df.groupby('Discount_Tier', observed=False).agg(
    Transactions=('Transaction_ID', 'count'),
    Avg_Quantity=('Quantity', 'mean'),
    Avg_Gross_Sales=('Gross_Sales', 'mean'),
    Avg_Net_Sales=('Total_Amount', 'mean')
).round(2)

print("--- Impact of Discount Tiers on Basket Size & Net Revenue ---")
print(discount_impact)

# 2. Correlation Analysis using Pandas / NumPy
corr_columns = ['Discount_Pct', 'Quantity', 'Gross_Sales', 'Total_Amount']
correlation_matrix = df[corr_columns].corr().round(3)

print("\n--- Correlation Matrix ---")
print(correlation_matrix)

# Interpretation
qty_corr = correlation_matrix.loc['Discount_Pct', 'Quantity']
net_corr = correlation_matrix.loc['Discount_Pct', 'Total_Amount']
print(f"\nCorrelation between Discount % and Quantity   : {qty_corr}")
print(f"Correlation between Discount % and Total Amount: {net_corr}")
print("Finding: Higher discounts do NOT lead to significantly higher units purchased, but they directly erode net revenue.")
```

```python

# Question 7: Which customers appear to make frequent or high-value purchases?


# Group by Customer
customer_summary = df.groupby('Customer_ID').agg(
    Total_Spent=('Total_Amount', 'sum'),
    Visits_Count=('Transaction_ID', 'count'),
    Avg_Spent_Per_Visit=('Total_Amount', 'mean'),
    Total_Items=('Quantity', 'sum')
).reset_index().round(2)

# Top 5 High-Value Customers (Highest Revenue Spenders)
top_spenders = customer_summary.sort_values(by='Total_Spent', ascending=False).head(5)

# Top 5 Frequent Shoppers (Highest Visit Frequency)
frequent_shoppers = customer_summary.sort_values(by='Visits_Count', ascending=False).head(5)

print("--- Top 5 High-Value Customers (Highest Total Spend) ---")
print(top_spenders)

print("\n--- Top 5 Frequent Shoppers (Most Transactions) ---")
print(frequent_shoppers)
```
